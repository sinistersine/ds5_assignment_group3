The Mandelbrot set is a set of complex numbers defined by the sequence:

with complex number 𝑐 = 𝑥 + 𝑦𝑖. The number 𝑐 is part of the Mandelbrot set when 𝑎𝑛 does not
diverge to infinity. In python, 𝑐 = 𝑐𝑜𝑚𝑝𝑙𝑒𝑥(𝑥,𝑦) is often used. This Mandelbrot set is a fractal

Normally, we would see if the sequence 𝑎𝑛 would diverge for 𝑛→∞, but we do not have enough
time to check it for infinite numbers. For this reason, we assume that the sequence diverges
to infinity if |𝑎𝑛| > 2 and we call this 𝑛 the diverging index. If we cannot find a diverging index
less than 100, then we assume that the sequence diverges to infinity and the diverging index
will be defined as 0.

```python
# Parameters die buiten de functie (in andere functies) worden gedefinieerd, zodat ze makkelijk kunnen worden aangepast. 
# type hinting voor overzicht
def mandelbrot_punt(c: complex, max_iter: int) -> int:
    """
    Bepaalt het aantal iteraties dat nodig is om te bepalen of een complex getal c behoort tot de Mandelbrot-verzameling.

    Parameters:
    c (complex): Het complexe getal dat wordt getest.
    max_iter (int): Het maximum aantal iteraties.

    Returns:
    int: Het aantal iteraties dat nodig is om te bepalen of c behoort tot de Mandelbrot-verzameling.
         Als het aantal iteraties gelijk is aan max_iter, behoort c tot de Mandelbrot-verzameling.
         

    Codeur: Anny
    """
    z = 0 # Bij beginpunt is z gelijk aan 0
    n = 0 # Het aantal iteraties wordt op 0 gezet, zodat het in de loop kan worden geteld
    for n in range(max_iter): # Zoek naar het aantal iteraties tot max_iter
        if abs(z) > 2: # Als de absolute waarde van z groter is dan 2, behoort c niet tot de Mandelbrot-verzameling
            # Oftewel, de ontsnapping van de Mandelbrot-verzameling is bereikt
            return n
        z = z*z + c # De gegeven formule in de opdracht, waarbij z wordt geüpdatet met de waarde van c die is gedefinieerd in de functie mandelbrot_set
    return max_iter # Moet in draw_mandel(max_iter) 

def mandelbrot_set(xmin: float, xmax: float, ymin: float, ymax: float, width: int, height: int, max_iter: int) -> np.ndarray:
    """
    Genereert een afbeelding van de Mandelbrot-verzameling.

    Parameters:
    xmin (float): De minimumwaarde van de x-as.
    xmax (float): De maximumwaarde van de x-as.
    ymin (float): De minimumwaarde van de y-as.
    ymax (float): De maximumwaarde van de y-as.
    width (int): Het aantal pixels in de breedte van de afbeelding.
    height (int): Het aantal pixels in de hoogte van de afbeelding.
    max_iter (int): Het maximum aantal iteraties.

    Returns:
    np.ndarray: Een 2D-array die het aantal iteraties voor elk punt in het complexe vlak bevat.
    
    Codeur: Anny
    """
    
    # Grenzen van het vlak doorgeven als parameters en het aantal pixels in de breedte en hoogte van de afbeelding.
    x = np.linspace(xmin, xmax, width) # Alleen met reële waarden, dus x-as
    y = np.linspace(ymin, ymax, height) # Alleen met imaginaire waarden, dus y-as, COMPLEX PLANE
    
    rij = []
    
       
    for a in y: # Loop door de imaginaire waarden
        rij_y = [] # Maak een lege lijst voor de rij
        
        for b in x: # Loop door de reële waarden
            c = complex(b, a) # Maak een complex getal van de reële en imaginaire waarden, c = b + ai
            iteraties = mandelbrot_punt(c, max_iter) # Bepaal het aantal iteraties voor het complex getal c
            rij_y.append(iteraties) # Voeg toe aan HUIDIGE rij
            
        rij.append(rij_y) # Voeg de rij toe aan de matrix van rijen
        print(f"Rij {len(rij)} van {height}") # Print voortgang van de berekening
    rij = np.array(rij) # Converteer de lijst van rijen naar een numpy array
    
    return rij # Return de numpy array van het aantal iteraties voor elk punt in het complexe vlak

# Mijn laptop crasht wanneer ik het probeer te visualiseren voor een reden dat nog niet duidelijk is, dus ik hoop dat jou wel lukt
# Als de loops korter kunnen, laat het weten

```
